In this problem, the aim is to aggregate the hourly temperature data for Helsinki Kumpula and Rovaniemi weather stations to the daily level. Currently, there are (at most) 3 measurements per hour in the data as you can see from the `YR--MODAHRMN` column (Year-Month-Day-Hour-Minute in Greenwich Mean Time (GMT):

```
    USAF  YR--MODAHRMN  TEMP  MAX  MIN  Celsius

1  28450  201705010020  30.0  NaN  NaN       -1
2  28450  201705010050  30.0  NaN  NaN       -1
3  28450  201705010100  31.0  NaN  NaN       -1
4  28450  201705010120  30.0  NaN  NaN       -1
```

The output should contain mean, max, and min Celsius temperatures for each day (for example, one mean temperature value for the 1st of May and so on).

````python
# Importa Pandas
import pandas as pd

# Define File Paths
fp_kumpula = "Kumpula_temps_May_Aug_2017.csv"
fp_rovaniemi = "Rovaniemi_temps_May_Aug_2017.csv"

# Read/Import the file
kumpula = pd.read_csv(fp_kumpula)
rovaniemi = pd.read_csv(fp_rovaniemi)

# Check the files Beginning and ending
print(kumpula.head())
print(kumpula.tail())
print(rovaniemi.head())
print(rovaniemi.tail())
````

````python
# Delete Unnamed: 0 column
kumpula = kumpula.drop(columns=["Unnamed: 0"])
rovaniemi = rovaniemi.drop(columns=["Unnamed: 0"])

# Check the files Beginning and ending
print(kumpula.head())
print(rovaniemi.head())
````

````python
# For each day I need to get the mean, max and min value.
# Resample dataframes into daily data Frames.

# Convert YR--MODAHRMN into date time
kumpula["YR--MODAHRMN"] = pd.to_datetime(kumpula["YR--MODAHRMN"], format = "%Y%m%d%H%M")
rovaniemi["YR--MODAHRMN"] = pd.to_datetime(kumpula["YR--MODAHRMN"], format = "%Y%m%d%H%M")

# Check the files Beginning and ending
print(kumpula.head())
print(rovaniemi.head())
````

````python
# Set the date column as index
kumpula = kumpula.set_index("YR--MODAHRMN")
rovaniemi = rovaniemi.set_index("YR--MODAHRMN")

# Check the files Beginning and ending
print(kumpula.head())
print(rovaniemi.head())
````

````python
#Check Celsius columns type
print(kumpula["Celsius"].dtype)
print(rovaniemi["Celsius"].dtype)
````

````python
#Resample into daily
kumpula_daily = kumpula["Celsius"].resample("d").agg(["mean", "max", "min"])
rovaniemi_daily = rovaniemi["Celsius"].resample("d").agg(["mean", "max", "min"])

# Check the files Beginning and ending
print(kumpula_daily.head())
print("")
print(rovaniemi_daily.head())
````

````python
#Renaming
kumpula_daily.index_name = "Date"

kumpula_daily = kumpula_daily.rename(
    columns={
        "mean": "Mean (ºC)",
        "max": "Max (ºC)",
        "min": "Min (ºC)"})

rovaniemi_daily.index_name = "Date"

rovaniemi_daily = rovaniemi_daily.rename
````

````python
# Save new csv.
output_kumpula_daily = "Kumpula_daily_temps_May_Aug_2017.csv"
output_rovaniemi_daily = "Rovaniemi_daily_temps_May_Aug_2017.csv"

kumpula_daily.to_csv(output_kumpula_daily, sep=",", float_format="%.2f")
rovaniemi_daily.to_csv(output_rovaniemi_daily, sep=",", float_format="%.2f")
````
