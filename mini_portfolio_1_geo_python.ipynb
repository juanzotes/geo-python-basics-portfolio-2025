## 1. Temperature Classification (with explanations)

````python
# List of temperature values
temperatures = [-5.4, 1.0, -1.3, -4.8, 3.9, 0.1, -4.4, 4.0, -2.2, -3.9]

# Empty lists to store categorized temperatures
cold = []          # t < -2
slippery = []      # -2 <= t < 2
comfortable = []   # 2 <= t < 15
warm = []          # t >= 15

# Loop through each temperature and classify it
for t in temperatures:
    if t < -2:
        cold.append(t)
    elif -2 <= t < 2:
        slippery.append(t)
    elif 2 <= t < 15:
        comfortable.append(t)
    else:
        warm.append(t)

# Display the categorized lists
cold, slippery, comfortable, warm
````

### Percentages of each category

````python
# Create a dictionary to iterate easily over categories
categories = {
    "cold": cold,
    "slippery": slippery,
    "comfortable": comfortable,
    "warm": warm
}

total = len(temperatures)

# Print the percentage of each category
for name, group in categories.items():
    percentage = (len(group) / total) * 100
    print(f"{name}: {percentage:.0f}%")
````

## 2. Coordinate Quadrant Classification (with explanations)

````python
# Lists representing latitudes, longitudes, and station names
lats = [60.1, 62.5, 58.9, 63.2]
lons = [24.9, 25.3, 21.0, 26.1]
stations = ["A", "B", "C", "D"]

# Cutoff values to create four quadrants
north_south_cutoff = 61.0
east_west_cutoff = 24.0

# Empty lists to store station classifications
north_west = []
north_east = []
south_west = []
south_east = []

# Enumerate gives index + station name
for n, station in enumerate(stations):
    lat = lats[n]   # get latitude using index
    lon = lons[n]   # get longitude using index

    # Determine quadrant
    if lat > north_south_cutoff and lon < east_west_cutoff:
        north_west.append(station)
    elif lat > north_south_cutoff and lon > east_west_cutoff:
        north_east.append(station)
    elif lat < north_south_cutoff and lon < east_west_cutoff:
        south_west.append(station)
    else:
        south_east.append(station)

north_west, north_east, south_west, south_east
````

## 3. Functions, Inverse Functions, and Combining Functions


This section explains how functions work, how to combine them, and whether inverse functions can be created automatically.

### Key ideas:
- A **function** is defined with `def` and returns a value.
- You **cannot** automatically get an inverse function in Python.
- To get an inverse, you must **define a new function** manually.
- You can **combine functions** by calling one inside another.


````python
# Basic function example
def celsius_to_fahrenheit(c):
    """Convert Celsius to Fahrenheit."""
    return (c * 9/5) + 32

# Combine functions: Fahrenheit → Celsius → Fahrenheit
def fahrenheit_to_celsius(f):
    """Convert Fahrenheit to Celsius."""
    return (f - 32) * 5/9

# Demonstrate inverse-like behavior
c = 10
f = celsius_to_fahrenheit(c)
c_recovered = fahrenheit_to_celsius(f)

c, f, c_recovered
````

### ⚠️ Can Python create inverse functions automatically?
**No.**  
Python does not compute the inverse of a function on its own.  

If you want an inverse, you must **define it yourself**, like we did with:

- `celsius_to_fahrenheit(c)`
- `fahrenheit_to_celsius(f)` (inverse manually created)

### ✔️ Combining functions
You can call a function inside another:

```python
def double(x):
    return x * 2

def double_then_add_three(x):
    return double(x) + 3
```

The second function uses the first function internally.


````python
def double(x):
    return x * 2

def double_then_add_three(x):
    return double(x) + 3
````

````python
print(double(4))
print(double_then_add_three(4))
````
